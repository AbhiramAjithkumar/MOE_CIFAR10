```python
import random  # Provides random sampling for selecting examples.
from torch.utils.data import DataLoader, Subset  # Imports tools for making subsets and batches.

random.seed(42)  # Makes the random selection repeatable.
train_indices = random.sample(range(len(train_full)), 5000)  # Selects 5,000 training-image indexes.
test_indices = random.sample(range(len(test_full)), 1000)  # Selects 1,000 test-image indexes.

train_data = Subset(train_full, train_indices)  # Keeps only the selected training images.
test_data = Subset(test_full, test_indices)  # Keeps only the selected test images.

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)  # Makes shuffled training batches of 64 images.
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)  # Makes test batches of 64 without shuffling.

images, labels = next(iter(train_loader))  # Reads the first batch to check that the loader works.
print("Image batch shape:", images.shape)  # Shows the batch dimensions.
print("Label batch shape:", labels.shape)  # Shows the number of labels in the batch.
print("First few labels:", [train_full.classes[i] for i in labels[:8]])  # Displays the first 8 category names.
```

```python
class CNNFeatureExtractor(nn.Module):  # Defines a CNN that converts images into feature vectors.
    def __init__(self, feature_dim=256):  # Sets up the layers; the final feature vector has 256 values.
        super().__init__()  # Initializes the PyTorch neural-network base class.
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)  # Finds simple patterns in the 3 color channels.
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)  # Builds more detailed patterns from the first layer.
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)  # Builds higher-level image features.
        self.fc = nn.Linear(128 * 4 * 4, feature_dim)  # Converts the final feature maps into a 256-value vector.

    def forward(self, x):  # Describes how an image passes through the layers.
        x = F.relu(self.conv1(x))  # Applies the first convolution and keeps positive activations.
        x = F.max_pool2d(x, 2)  # Halves image width and height from 32×32 to 16×16.
        x = F.relu(self.conv2(x))  # Applies the second convolution.
        x = F.max_pool2d(x, 2)  # Halves width and height from 16×16 to 8×8.
        x = F.relu(self.conv3(x))  # Applies the third convolution.
        x = F.max_pool2d(x, 2)  # Halves width and height from 8×8 to 4×4.
        x = torch.flatten(x, start_dim=1)  # Flattens each image's 128 feature maps into 2,048 values.
        x = self.fc(x)  # Maps those 2,048 values to the 256-value feature vector.
        return x  # Returns the extracted features.

cnn = CNNFeatureExtractor().to(device)  # Creates the CNN and places it on the GPU if available.
sample_images = images.to(device)  # Moves the example image batch to the same device as the CNN.
features = cnn(sample_images)  # Runs the example images through the CNN.
print("Feature batch shape:", features.shape)  # Checks that we get 64 feature vectors with 256 values each.
```

```python
class GatingNetwork(nn.Module):  # Defines the network that chooses which experts to use.
    def __init__(self, input_dim=256, num_experts=5, top_k=2):  # Sets the feature size, expert count, and number selected.
        super().__init__()  # Initializes the PyTorch neural-network base class.
        self.router = nn.Linear(input_dim, num_experts)  # Scores each expert using the image's 256 features.
        self.top_k = top_k  # Stores how many experts the gate will select.

    def forward(self, features):  # Takes a batch of CNN feature vectors as input.
        logits = self.router(features)  # Produces one raw score per expert for each image.
        top_values, top_indices = torch.topk(logits, self.top_k, dim=1)  # Finds the two highest-scoring experts per image.
        selected_weights = F.softmax(top_values, dim=1)  # Converts their scores into weights that sum to 1.
        gate_weights = torch.zeros_like(logits).scatter(1, top_indices, selected_weights)  # Places those weights in a five-expert vector; other experts get 0.
        return gate_weights  # Returns one set of expert weights per image.

gate = GatingNetwork().to(device)  # Creates the gate and places it on the same device as the CNN.
gate_weights = gate(features)  # Gets expert weights for the feature vectors from the previous cell.
print("Gate output shape:", gate_weights.shape)  # Shows one weight for each of the 5 experts per image.
print("First image's expert weights:", gate_weights[0].detach().cpu())  # Displays the weights for the first image.
print("Number of selected experts:", torch.count_nonzero(gate_weights[0]).item())  # Confirms that 2 experts were selected.
print("Weights sum to:", gate_weights[0].sum().item())  # Confirms that the selected weights add up to 1.
```

```python
FEATURE_DIM = 256  # Sets how many values each image's feature vector contains.
HIDDEN_DIM = 128  # Sets the size of each expert's internal layer.
NUM_EXPERTS = 5  # Keeps the expert count consistent with the gate.

class Expert(nn.Module):  # Defines one small neural network that processes image features.
    def __init__(self, input_dim=FEATURE_DIM, hidden_dim=HIDDEN_DIM, output_dim=FEATURE_DIM):  # Sets the sizes of its layers.
        super().__init__()  # Initializes the PyTorch neural-network base class.
        self.fc1 = nn.Linear(input_dim, hidden_dim)  # Maps 256 input features to 128 hidden values.
        self.fc2 = nn.Linear(hidden_dim, output_dim)  # Maps the 128 hidden values back to 256 values.

    def forward(self, x):  # Describes how input features pass through this expert.
        x = F.relu(self.fc1(x))  # Applies the first layer and keeps positive activations.
        x = self.fc2(x)  # Produces the expert's 256-value output.
        return x  # Returns the processed feature vector.

experts = nn.ModuleList([Expert() for _ in range(NUM_EXPERTS)]).to(device)  # Creates five independently initialized experts.
expert_outputs = torch.stack([expert(features) for expert in experts], dim=1)  # Runs every image through every expert and stacks the outputs.
print("All expert outputs shape:", expert_outputs.shape)  # Shows batch size, expert count, and output size.
```

```python
weighted_outputs = gate_weights.unsqueeze(-1) * expert_outputs  # Applies each image's expert weights to its expert outputs.
combined_features = weighted_outputs.sum(dim=1)  # Adds the five weighted outputs into one vector per image.

print("Combined feature shape:", combined_features.shape)  # Shows the single combined vector produced for each image.
```

```python
NUM_CLASSES = 10  # CIFAR-10 has 10 categories.

classifier = nn.Linear(FEATURE_DIM, NUM_CLASSES).to(device)  # Maps each 256-value MoE output to 10 category scores.
logits = classifier(combined_features)  # Calculates the 10 raw scores for each image.
predicted_ids = logits.argmax(dim=1)  # Picks the highest-scoring category for each image.

print("Logits shape:", logits.shape)  # Checks that we have 64 images and 10 scores per image.
print("First predicted category:", train_full.classes[predicted_ids[0].item()])  # Shows the first image's current prediction.
print("Correct category:", train_full.classes[labels[0].item()])  # Shows the correct label so we can compare.
```

```python
class CNNMoE(nn.Module):  # Combines the CNN, MoE gate, experts, and final classifier.
    def __init__(self):  # Creates all learnable parts of the full model.
        super().__init__()  # Initializes the PyTorch neural-network base class.
        self.cnn = CNNFeatureExtractor(feature_dim=FEATURE_DIM)  # Converts each image into a 256-value feature vector.
        self.gate = GatingNetwork(input_dim=FEATURE_DIM, num_experts=NUM_EXPERTS, top_k=2)  # Selects two experts per image.
        self.experts = nn.ModuleList([Expert() for _ in range(NUM_EXPERTS)])  # Creates five independent expert networks.
        self.classifier = nn.Linear(FEATURE_DIM, NUM_CLASSES)  # Converts the combined expert output into 10 class scores.

    def forward(self, images):  # Defines how a batch of images flows through the full model.
        features = self.cnn(images)  # Extracts a 256-value feature vector from each image.
        gate_weights = self.gate(features)  # Calculates each expert's weight for each image.
        expert_outputs = torch.stack([expert(features) for expert in self.experts], dim=1)  # Runs every image through all five experts.
        combined_features = (gate_weights.unsqueeze(-1) * expert_outputs).sum(dim=1)  # Blends the expert outputs using the gate weights.
        logits = self.classifier(combined_features)  # Produces 10 raw class scores per image.
        return logits, gate_weights  # Returns class scores and gate choices for training and inspection.

model = CNNMoE().to(device)  # Creates the complete model and places it on the T4 GPU.
sample_logits, sample_gate_weights = model(sample_images)  # Runs the sample batch through the complete model.
print("Class-score shape:", sample_logits.shape)  # Checks for 64 images with 10 class scores each.
print("Gate-weight shape:", sample_gate_weights.shape)  # Checks for 64 images with 5 expert weights each.
```

```python
criterion = nn.CrossEntropyLoss()  # Measures how far the 10 class scores are from the correct labels.
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)  # Updates all model weights using the Adam algorithm.
EPOCHS = 7  # Sets a short initial run; we can increase it after confirming everything works.
BALANCE_COEF = 0.001  # Sets how strongly the auxiliary expert-balancing term affects training.

for epoch in range(EPOCHS):  # Repeats training over the small training dataset.
    model.train()  # Turns on training behavior for the model.
    running_loss = 0.0  # Accumulates the loss over batches in this epoch.
    correct = 0  # Counts correctly predicted training images.
    total = 0  # Counts training images processed in this epoch.

    for batch_images, batch_labels in train_loader:  # Gets the next batch of images and their correct categories.
        batch_images = batch_images.to(device)  # Sends images to the T4 GPU.
        batch_labels = batch_labels.to(device)  # Sends labels to the same GPU.

        optimizer.zero_grad()  # Clears gradients left over from the previous batch.
        logits, gate_weights = model(batch_images)  # Gets category scores and expert weights for this batch.
        classification_loss = criterion(logits, batch_labels)  # Measures the model's category-prediction error.
        expert_usage = gate_weights.mean(dim=0)  # Averages each expert's gate weight across this batch.
        balance_loss = (expert_usage * torch.log(expert_usage + 1e-8)).sum()  # Encourages gate weight to be shared among experts.
        loss = classification_loss + BALANCE_COEF * balance_loss  # Combines prediction error with the small balancing term.
        loss.backward()  # Calculates how each learnable weight contributed to the loss.
        optimizer.step()  # Updates the model weights using those gradients.

        running_loss += loss.item()  # Adds this batch's loss to the epoch total.
        predictions = logits.argmax(dim=1)  # Chooses the highest-scoring category for each image.
        correct += (predictions == batch_labels).sum().item()  # Adds the number of correct predictions in this batch.
        total += batch_labels.size(0)  # Adds the number of images in this batch.

    average_loss = running_loss / len(train_loader)  # Calculates the average loss across the training batches.
    train_accuracy = correct / total  # Calculates the fraction of training images classified correctly.
    print(f"Epoch {epoch + 1}/{EPOCHS} | loss: {average_loss:.4f} | training accuracy: {train_accuracy:.1%}")  # Reports progress for this epoch.
```

```python
model.eval()  # Switches the model to evaluation behavior.
test_correct = 0  # Counts correct predictions on the test images.
test_total = 0  # Counts how many test images we evaluate.
test_loss_total = 0.0  # Accumulates prediction loss across test batches.
expert_weight_totals = torch.zeros(NUM_EXPERTS, device=device)  # Stores the total gate weight received by each expert.

with torch.no_grad():  # Disables gradient calculations because we are only evaluating.
    for batch_images, batch_labels in test_loader:  # Gets the next batch of test images and labels.
        batch_images = batch_images.to(device)  # Sends test images to the T4 GPU.
        batch_labels = batch_labels.to(device)  # Sends test labels to the same GPU.
        logits, gate_weights = model(batch_images)  # Gets class scores and expert weights from the model.
        loss = criterion(logits, batch_labels)  # Measures prediction error for this test batch.
        predictions = logits.argmax(dim=1)  # Chooses the highest-scoring category for each test image.
        test_correct += (predictions == batch_labels).sum().item()  # Counts correct predictions in this batch.
        test_total += batch_labels.size(0)  # Counts the test images in this batch.
        test_loss_total += loss.item() * batch_labels.size(0)  # Adds loss weighted by the number of images.
        expert_weight_totals += gate_weights.sum(dim=0)  # Adds this batch's routing weights to each expert's total.

test_accuracy = test_correct / test_total  # Calculates the fraction of test images classified correctly.
average_test_loss = test_loss_total / test_total  # Calculates the average test loss per image.
average_expert_weights = expert_weight_totals / test_total  # Calculates each expert's average gate weight per image.

print(f"Test accuracy: {test_accuracy:.1%}")  # Reports accuracy on images not used for training.
print(f"Test loss: {average_test_loss:.4f}")  # Reports average prediction error on the test set.
print("Average gate weight per expert:", average_expert_weights.cpu().tolist())  # Shows how routing weight was shared across experts.
```
